```python
import json

# ====================== Configuration ======================
INPUT_JSON = "batch_sticker_prompts.json"
OUTPUT_JSON = "batch_sticker_prompts_baseline.json"

# Load the original JSON file
with open(INPUT_JSON, "r", encoding="utf-8") as f:
    data = json.load(f)

# Remove all occurrences of "sks" in each prompt
for item in data:
    if "sd_sticker_prompt" in item:
        prompt = item["sd_sticker_prompt"]
        # Clean up "sks" and extra spaces
        prompt = prompt.replace("sks ", " ").replace(" sks", " ").replace("sks", "")
        # Remove duplicate spaces caused by replacement
        prompt = " ".join(prompt.split())
        item["sd_sticker_prompt"] = prompt

# Save the cleaned result to a new JSON file
with open(OUTPUT_JSON, "w", encoding="utf-8") as f:
    json.dump(data, f, indent=4, ensure_ascii=False)

print("✅ All 'sks' tokens have been removed.")
print(f"✅ Cleaned file saved as: {OUTPUT_JSON}")
```
